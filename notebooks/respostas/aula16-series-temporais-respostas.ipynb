## 🎯 A média móvel

```python
def media_movel(serie, janela):
    n = len(serie) - janela + 1
    medias = np.zeros(n)
    for i in range(n):
        medias[i] = np.mean(serie[i:i + janela])
    return medias
```

## 🎯 O erro absoluto médio

```python
def mae(previsto, real):
    return np.mean(abs(previsto - real))
```

## 🎯 O pico da próxima semana

```python
def pico_da_proxima_semana(serie):
    n = len(serie)
    coef = np.polyfit(np.linspace(0, n - 1, n), serie, 1)
    previsao = serie[n - 168:] + coef[0] * 168
    k = np.argmax(previsao)
    return (np.max(previsao), k)
```
