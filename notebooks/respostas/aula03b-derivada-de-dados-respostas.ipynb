## 🎯 Velocidade num ponto

```python
def velocidade_kmh(t, x, i):
    v = (x[i + 1] - x[i - 1]) / (t[i + 1] - t[i - 1])
    return v * 3.6
```

## 🎯 A maior taxa

```python
def maior_taxa(t, y, k):
    maior = (y[2 * k] - y[0]) / (t[2 * k] - t[0])
    for i in range(k, len(t) - k):
        taxa = (y[i + k] - y[i - k]) / (t[i + k] - t[i - k])
        if taxa > maior:
            maior = taxa
    return maior
```

## 🎯 O primeiro alerta

```python
def primeiro_alerta(t, nivel, k, limite):
    for i in range(k, len(t) - k):
        taxa = (nivel[i + k] - nivel[i - k]) / (t[i + k] - t[i - k])
        if taxa > limite:
            return t[i]
    return -1
```
