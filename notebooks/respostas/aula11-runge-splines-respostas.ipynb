## 🎯 Um nó de Chebyshev

```python
def no_chebyshev(k, n):
    return np.cos((2 * k + 1) * np.pi / (2 * n))
```

## 🎯 Tem bico?

```python
def bico(xs, ys, i):
    esquerda = (ys[i] - ys[i - 1]) / (xs[i] - xs[i - 1])
    direita = (ys[i + 1] - ys[i]) / (xs[i + 1] - xs[i])
    return direita - esquerda
```

## 🎯 O voo é seguro?

```python
def voo_seguro(alturas, minima, maxima):
    for h in alturas:
        if h < minima or h > maxima:
            return False
    return True
```
