## 🎯 Quantas vezes até chegar lá?

```python
def vezes_ate(valor, alvo):
    total = 0.0
    vezes = 0
    while total < alvo:
        total = total + valor
        vezes = vezes + 1
    return vezes
```

## 🎯 O erro relativo aproximado

```python
def eps_a(atual, anterior):
    return abs((atual - anterior) / atual) * 100
```
