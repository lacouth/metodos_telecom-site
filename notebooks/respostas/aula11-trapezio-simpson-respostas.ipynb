## 🎯 A regra do trapézio

```python
def trapezio(f, a, b, n):
    h = (b - a) / n
    y = f(np.linspace(a, b, n + 1))
    return h / 2 * (y[0] + 2 * np.sum(y[1:-1]) + y[-1])
```

## 🎯 A regra de Simpson

```python
def simpson(f, a, b, n):
    h = (b - a) / n
    y = f(np.linspace(a, b, n + 1))
    return h / 3 * (y[0] + 4 * np.sum(y[1:-1:2]) + 2 * np.sum(y[2:-1:2]) + y[-1])
```

## 🎯 A área de uma tabela qualquer

```python
def trapezio_dados(x, y):
    total = 0.0
    for i in range(len(x) - 1):
        total = total + (x[i + 1] - x[i]) * (y[i] + y[i + 1]) / 2
    return total
```
