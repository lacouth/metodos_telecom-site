## 🎯 Uma iteração de Jacobi

```python
def iteracao_jacobi(x1, x2, x3):
    novo1 = (7.85 + 0.1 * x2 + 0.2 * x3) / 3
    novo2 = (-19.3 - 0.1 * x1 + 0.3 * x3) / 7
    novo3 = (71.4 - 0.3 * x1 + 0.2 * x2) / 10
    return (novo1, novo2, novo3)
```

## 🎯 O valor novo de uma linha

```python
def valor_novo(A, b, x, i):
    soma = b[i]
    for j in range(len(b)):
        if j != i:
            soma = soma - A[i, j] * x[j]
    return soma / A[i, i]
```

## 🎯 Tem dominância?

```python
def diagonal_dominante(A):
    n = len(A)
    for i in range(n):
        soma = 0.0
        for j in range(n):
            if j != i:
                soma = soma + abs(A[i, j])
        if abs(A[i, i]) <= soma:
            return False
    return True
```
