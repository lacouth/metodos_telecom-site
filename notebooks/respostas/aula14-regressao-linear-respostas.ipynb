## 🎯 Os resíduos

```python
def residuos(x, y, a0, a1):
    return y - (a0 + a1 * x)
```

## 🎯 O espalhamento em volta da reta

```python
def desvio_residuos(x, y, a0, a1):
    e = y - (a0 + a1 * x)
    return np.sqrt(np.sum(e**2) / (len(x) - 2))
```

## 🎯 O ano do limite

```python
def ano_do_limite(anos, valores, limite):
    coef = np.polyfit(anos, valores, 1)
    return (limite - coef[1]) / coef[0]
```
