## 🎯 Deu triangular?

```python
def eh_triangular(U):
    n = len(U)
    for i in range(n):
        for j in range(i):
            if abs(U[i, j]) > 1e-12:
                return False
    return True
```

## 🎯 Quem deve ser o pivô?

```python
def linha_do_pivo(A, k):
    p = k
    for i in range(k + 1, len(A)):
        if abs(A[i, k]) > abs(A[p, k]):
            p = i
    return p
```

## 🎯 O resíduo

```python
def residuo_maximo(A, x, b):
    r = b - A @ x
    maior = 0.0
    for valor in r:
        if abs(valor) > maior:
            maior = abs(valor)
    return maior
```

## 🎯 A posição do celular

```python
def posicao(torres, distancias):
    x1, y1 = torres[0]
    r1 = distancias[0]
    A = np.zeros((2, 2))
    b = np.zeros(2)
    for k in range(2):
        xk, yk = torres[k + 1]
        rk = distancias[k + 1]
        A[k, 0] = 2 * (xk - x1)
        A[k, 1] = 2 * (yk - y1)
        b[k] = r1**2 - rk**2 + xk**2 - x1**2 + yk**2 - y1**2
    return np.linalg.solve(A, b)
```
