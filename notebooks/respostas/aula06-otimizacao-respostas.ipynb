## 🎯 Um passo de Newton para otimização

```python
def passo_otimizacao(f, x):
    h = 1e-4
    d1 = (f(x + h) - f(x - h)) / (2 * h)
    d2 = (f(x + h) - 2 * f(x) + f(x - h)) / h**2
    return x - d1 / d2
```

## 🎯 Vale ou morro?

```python
def tipo_de_ponto(f, x):
    h = 1e-4
    d2 = (f(x + h) - 2 * f(x) + f(x - h)) / h**2
    if d2 > 0:
        return "minimo"
    return "maximo"
```

## 🎯 O máximo na grade

```python
def maximo_na_grade(f, a, b, n):
    xs = np.linspace(a, b, n)
    return xs[np.argmax(f(xs))]
```
