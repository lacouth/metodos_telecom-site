## 🎯 A função central

```python
def central(f, x, h):
    return (f(x + h) - f(x - h)) / (2 * h)
```

## 🎯 Acelerando ou freando?

```python
def fase(t):
    h = 0.01
    d2 = (P(t + h) - 2 * P(t) + P(t - h)) / h**2
    if d2 > 0:
        return "acelerando"
    return "freando"
```

## 🎯 A regressiva

```python
def regressiva(f, x, h):
    return (f(x) - f(x - h)) / h
```

## 🎯 O café esfriando

```python
def taxa_cafe(t):
    h = 0.01
    return (T(t + h) - T(t - h)) / (2 * h)
```

## 🎯 A maré mais rápida

```python
def maior_subida(horas):
    h = 0.01
    melhor_hora = horas[0]
    maior_taxa = (mare(horas[0] + h) - mare(horas[0] - h)) / (2 * h)
    for t in horas:
        taxa = (mare(t + h) - mare(t - h)) / (2 * h)
        if taxa > maior_taxa:
            maior_taxa = taxa
            melhor_hora = t
    return (melhor_hora, maior_taxa * 100)
```
