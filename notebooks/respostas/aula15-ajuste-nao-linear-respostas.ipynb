## 🎯 Avaliar o polinômio do polyfit

```python
def avalia_polinomio(coef, x):
    grau = len(coef) - 1
    total = 0.0
    for k in range(len(coef)):
        total = total + coef[k] * x**(grau - k)
    return total
```

## 🎯 O modelo de propagação

```python
def modelo_propagacao(distancia, rssi):
    coef = np.polyfit(np.log10(distancia), rssi, 1)
    A = coef[1]
    n = -coef[0] / 10
    return (A, n)
```
