## 🎯 Um passo de Newton

```python
def passo_newton(f, df, x):
    return x - f(x) / df(x)
```

## 🎯 O ponto da secante

```python
def ponto_secante(f, x_ant, x):
    return x - f(x) * (x_ant - x) / (f(x_ant) - f(x))
```

## 🎯 A raiz quadrada da calculadora

```python
def raiz_quadrada(a):
    x = a
    for it in range(100):
        novo = (x + a / x) / 2
        if abs(novo - x) < 1e-12 * novo:
            return novo
        x = novo
    return x
```
