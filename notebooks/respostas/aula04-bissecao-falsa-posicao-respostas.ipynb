## 🎯 Um passo da bisseção

```python
def um_passo(f, a, b):
    xm = (a + b) / 2
    if f(a) * f(xm) < 0:
        return (a, xm)
    return (xm, b)
```

## 🎯 O ponto da falsa posição

```python
def ponto_falsa_posicao(f, a, b):
    return b - f(b) * (a - b) / (f(a) - f(b))
```

## 🎯 O VPL de qualquer investimento

```python
def vpl(r, fluxos):
    total = 0
    for ano in range(len(fluxos)):
        total = total + fluxos[ano] / (1 + r)**ano
    return total
```

## 🎯 A bisseção como função

```python
def bissecao(f, a, b, tol):
    xm = a
    for it in range(100):
        anterior = xm
        xm = (a + b) / 2
        if f(a) * f(xm) < 0:
            b = xm
        else:
            a = xm
        if abs((xm - anterior) / xm) * 100 < tol:
            break
    return xm
```
