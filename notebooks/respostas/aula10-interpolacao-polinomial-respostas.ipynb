## 🎯 A reta entre dois pontos

```python
def reta_entre(x0, y0, x1, y1, x):
    return y0 + (y1 - y0) / (x1 - x0) * (x - x0)
```

## 🎯 Uma peça da base

```python
def base_lagrange(xs, i, x):
    L = 1.0
    for j in range(len(xs)):
        if j != i:
            L = L * (x - xs[j]) / (xs[i] - xs[j])
    return L
```

## 🎯 Quanto o polinômio erra?

```python
def erro_maximo(xs, ys, f, pontos):
    maior = 0.0
    for x in pontos:
        erro = abs(lagrange(xs, ys, x) - f(x))
        if erro > maior:
            maior = erro
    return maior
```
