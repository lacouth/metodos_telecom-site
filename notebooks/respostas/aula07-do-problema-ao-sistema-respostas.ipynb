## 🎯 A mistura no laboratório

```python
def mistura(total, porcento):
    A = np.array([[1.0, 1.0, 1.0],          # volume
                  [0.20, 0.35, 0.50],       # ácido
                  [-2.0, 1.0, 0.0]])        # B = 2A
    b = np.array([total, porcento / 100 * total, 0.0])
    return np.linalg.solve(A, b)
```

## 🎯 As correntes de malha

```python
def malhas(v1, v3):
    A = np.array([[15.0, -5.0, 0.0],
                  [-5.0, 20.0, -10.0],
                  [0.0, -10.0, 25.0]])
    b = np.array([v1, 0.0, v3])
    return np.linalg.solve(A, b)
```

## 🎯 As ruas sem sensor

```python
def fluxos(x4):
    A = np.array([[1.0, 0.0, 0.0, -1.0],     # cruzamento A: x1 - x4 = 400 - 150
                  [-1.0, 1.0, 0.0, 0.0],     # cruzamento B: x2 - x1 = 300 - 250
                  [0.0, -1.0, 1.0, 0.0],     # cruzamento C: x3 - x2 = 200 - 350
                  [0.0, 0.0, 0.0, 1.0]])     # o sensor: x4 medido
    b = np.array([250.0, 50.0, -150.0, x4])
    return np.linalg.solve(A, b)
```
