## 🎯 Quando passa de 100 km/h?

```python
def instante_100kmh(dt):
    g = 9.81
    m = 68.1
    c = 0.25
    v = 0.0
    t = 0.0
    while v * 3.6 < 100:
        v = v + (g - (c / m) * v**2) * dt
        t = t + dt
    return t
```

## 🎯 A tensão chega a 90 %?

```python
def tempo_ate(fracao, dt):
    Vf = 5.0
    RC = 1.0
    V = 0.0
    t = 0.0
    while V < fracao * Vf:
        V = V + (Vf - V) / RC * dt
        t = t + dt
    return t
```

## 🎯 O paraquedas aberto

```python
def velocidade_final(m, c, v0, dt, t_final):
    g = 9.81
    v = v0
    for i in range(round(t_final / dt)):
        v = v + (g - (c / m) * v**2) * dt
    return v
```
