## 🎯 A sonda

```python
def amplitude_em(sinal, k):
    N = len(sinal)
    soma_cos = 0.0
    soma_sen = 0.0
    for n in range(N):
        soma_cos = soma_cos + sinal[n] * np.cos(2 * np.pi * k * n / N)
        soma_sen = soma_sen + sinal[n] * np.sin(2 * np.pi * k * n / N)
    return 2 * np.sqrt(soma_cos ** 2 + soma_sen ** 2) / N
```

## 🎯 A frequência dominante

```python
def frequencia_dominante(sinal, fs):
    variacao = sinal - np.mean(sinal)
    amplitude = np.abs(np.fft.rfft(variacao))
    freq = np.fft.rfftfreq(len(sinal), 1 / fs)
    return freq[np.argmax(amplitude)]
```

## 🎯 Os picos do espectro

```python
def picos(sinal, fs, limiar):
    N = len(sinal)
    amplitude = 2 * np.abs(np.fft.rfft(sinal - np.mean(sinal))) / N
    freq = np.fft.rfftfreq(N, 1 / fs)
    achados = []
    for k in range(len(freq)):
        if amplitude[k] > limiar:
            achados.append(freq[k])
    return achados
```
