## 🎯 A regressiva no fim

```python
def ultima_taxa(t, y):
    n = len(t)
    return (y[n - 1] - y[n - 2]) / (t[n - 1] - t[n - 2])
```

## 🎯 Casos novos na semana

```python
def novos_na_semana(acumulados, dia):
    return acumulados[dia] - acumulados[dia - 7]
```
