## 🎯 A área de uma faixa

```python
def faixa(t0, t1, y0, y1):
    return (t1 - t0) * (y0 + y1) / 2
```

## 🎯 A integral acumulada

```python
def acumulada(t, y):
    E = np.zeros(len(t))
    for i in range(1, len(t)):
        E[i] = E[i - 1] + (t[i] - t[i - 1]) * (y[i] + y[i - 1]) / 2
    return E
```

## 🎯 Compra e venda

```python
def energia_da_rede(t, geracao, consumo):
    compra = np.zeros(len(t))
    venda = np.zeros(len(t))
    for i in range(len(t)):
        saldo = geracao[i] - consumo[i]
        if saldo > 0:
            venda[i] = saldo
        else:
            compra[i] = -saldo
    total_compra = 0.0
    total_venda = 0.0
    for i in range(len(t) - 1):
        total_compra = total_compra + (t[i + 1] - t[i]) * (compra[i] + compra[i + 1]) / 2
        total_venda = total_venda + (t[i + 1] - t[i]) * (venda[i] + venda[i + 1]) / 2
    return (total_compra, total_venda)
```
